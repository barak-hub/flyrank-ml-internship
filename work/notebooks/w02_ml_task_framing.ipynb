I will use the **F1-score** as the primary evaluation metric for my binary classification task, supported by precision and recall.

Precision measures how many pages predicted to experience declining impressions actually decline. Recall measures how many pages with declining impressions the model successfully identifies.

F1-score balances these two measures. This matters because false positives could waste SEO specialists' time, while false negatives could cause important pages to be overlooked.

**Unit of Analysis: One Webpage**

Each row in the dataset represents one content item (webpage), identified by `content_id`.

The dataset contains 30,000 rows. For each content item, I selected historical Google Search impressions, clicks, and average ranking position.

I also created a binary target called `declining_impressions`, where 1 indicates fewer impressions in the latest 30-day period than in the previous 30-day period, and 0 indicates no decline.

This dataframe demonstrates how the classification problem can be represented using real data. It does not mean that a machine learning model has already been trained.

```python

import pandas as pd

url = "https://raw.githubusercontent.com/barak-hub/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

columns = [
    "content_id",
    "impressions_prev_30d",
    "impressions_last_30d",
    "clicks_prev_30d",
    "clicks_last_30d",
    "avg_position"
]

pages = df[columns].copy()

pages["declining_impressions"] = (
    pages["impressions_last_30d"] < pages["impressions_prev_30d"]
).astype(int)

print("Number of rows:", len(pages))
print("One row represents one content item (webpage).")

display(pages.head())
```

### Why Machine Learning Could Be Better Than a Fixed Rule

A fixed rule can identify pages whose Google Search impressions have already declined by comparing the latest 30 days with the previous 30 days.

However, this rule only detects a decline that has already happened. It does not predict which pages might experience declining impressions in the future.

Machine learning could potentially improve this process by learning patterns from historical information, such as previous impressions, clicks, ranking position, content age, and other relevant features.

A trained classification model could estimate which pages are at greater risk of future declines, allowing SEO specialists to prioritize investigation before performance worsens.

However, machine learning is not automatically better than a fixed rule. Its performance must be tested against a simple baseline using appropriate evaluation metrics and data from a later time period.

**Conclusion:** Machine learning would be useful only if it provides more reliable or actionable predictions than the existing rule. No model has been trained yet, so this remains a proposed approach rather than a proven improvement.
