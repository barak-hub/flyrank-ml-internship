```python
import pandas as pd

df = pd.read_parquet(file_path)

print("Number of rows:", len(df))

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())
```

```python
keys = ["client_hash_id", "content_hash_id", "report_date"]

print("Total rows:", len(df))
print("Unique client-content-date combinations:",
      df[keys].drop_duplicates().shape[0])

print("Duplicate combinations:",
      df.duplicated(subset=keys).sum())

print("Earliest date:", df["report_date"].min())
print("Latest date:", df["report_date"].max())

print("Unique clients:", df["client_hash_id"].nunique())
print("Unique webpages:", df["content_hash_id"].nunique())
```

```python
print("Total columns:", len(df.columns))

print("\nMissing values by column:")
print(
    df.isna()
      .sum()
      .sort_values(ascending=False)
      .to_string()
)
```

```python
print("GSC access:")
print(df["client_has_gsc"].value_counts(dropna=False))

print("\nGA4 access:")
print(df["client_has_ga4"].value_counts(dropna=False))

print("\nGSC average position availability:")
print(df["gsc_avg_position"].notna().value_counts())
```

```python
print("GSC-related columns:")
print([col for col in df.columns if col.startswith("gsc_")])

print("\nGSC data availability:")
print(df["gsc_data_available"].value_counts(dropna=False))
```

```python
available_df = df[df["gsc_data_available"].eq(True)]

print("Total rows:", len(df))
print("GSC available rows:", len(available_df))

print("\nMissing GSC metrics in available rows:")
print(
    available_df[
        ["gsc_impressions", "gsc_clicks", "gsc_avg_position"]
    ].isna().sum()
)
```

```python
print(
    available_df[
        ["gsc_impressions", "gsc_clicks", "gsc_avg_position"]
    ].describe().round(2)
)
```

```python
print("Daily GSC observations:")
print(
    available_df[
        ["report_date", "client_hash_id", "content_hash_id",
         "gsc_impressions", "gsc_clicks", "gsc_avg_position"]
    ].head(10).to_string(index=False)
)

print("\nUnique reporting dates:",
      available_df["report_date"].nunique())
```

```python
import pandas as pd

available_df = available_df.copy()

available_df["report_date"] = pd.to_datetime(
    available_df["report_date"]
)

print("Date type:", available_df["report_date"].dtype)
```

```python
history = available_df[
    available_df["report_date"].between(
        "2026-03-01", "2026-03-15"
    )
]

feature_df = (
    history.groupby(
        ["client_hash_id", "content_hash_id"]
    )
    .agg(
        historical_impressions=("gsc_impressions", "sum"),
        historical_clicks=("gsc_clicks", "sum"),
        historical_avg_position=("gsc_avg_position", "mean"),
        historical_ctr=("gsc_clicks", "sum"),
        impression_variability=("gsc_impressions", "std")
    )
    .reset_index()
)

feature_df["historical_ctr"] = (
    feature_df["historical_clicks"] /
    feature_df["historical_impressions"]
).fillna(0)

feature_df["impression_variability"] = (
    feature_df["impression_variability"].fillna(0)
)

print("Feature rows:", len(feature_df))
print(feature_df.head())
```

```python
import pandas as pd

available_df["report_date"] = pd.to_datetime(
    available_df["report_date"]
)

print("Date type:", available_df["report_date"].dtype)
```

```python
print("Feature rows:", len(feature_df))

print("\nMissing values:")
print(feature_df.isna().sum())

print("\nDuplicate client-page combinations:")
print(
    feature_df.duplicated(
        subset=["client_hash_id", "content_hash_id"]
    ).sum()
)
```

```python
print("History start:", history["report_date"].min())
print("History end:", history["report_date"].max())
print("Historical rows:", len(history))
print("Feature rows:", len(feature_df))
```

```python
future = available_df[
    available_df["report_date"].between(
        "2026-03-16", "2026-03-31"
    )
].copy()

print("Future start:", future["report_date"].min())
print("Future end:", future["report_date"].max())
print("Future rows:", len(future))
print("Unique client-page pairs:",
      future[["client_hash_id", "content_hash_id"]]
      .drop_duplicates().shape[0])
```

```python
future_pairs = future[
    ["client_hash_id", "content_hash_id"]
].drop_duplicates()

matched_pairs = feature_df.merge(
    future_pairs,
    on=["client_hash_id", "content_hash_id"],
    how="inner"
)

print("Historical feature pairs:", len(feature_df))
print("Future unique pairs:", len(future_pairs))
print("Matched pairs:", len(matched_pairs))

print(
    "Historical coverage (%):",
    round(len(matched_pairs) / len(feature_df) * 100, 2)
)
```

```python
future_targets = (
    future.groupby(
        ["client_hash_id", "content_hash_id"]
    )
    .agg(
        future_impressions=("gsc_impressions", "sum"),
        future_clicks=("gsc_clicks", "sum")
    )
    .reset_index()
)

print("Future target rows:", len(future_targets))

print("\nMissing values:")
print(future_targets.isna().sum())

print("\nFirst five rows:")
print(future_targets.head())
```

```python
model_df = feature_df.merge(
    future_targets,
    on=["client_hash_id", "content_hash_id"],
    how="inner",
    validate="one_to_one"
)

print("Model dataset rows:", len(model_df))

print("\nMissing values:")
print(model_df.isna().sum())

print("\nDuplicate client-page pairs:")
print(
    model_df.duplicated(
        subset=["client_hash_id", "content_hash_id"]
    ).sum()
)

print("\nFirst five rows:")
print(model_df.head())
```

```python
print("Future clicks summary:")
print(model_df["future_clicks"].describe())

print("\nPercentage with zero future clicks:")
print(
    round(
        (model_df["future_clicks"] == 0).mean() * 100,
        2
    )
)

print("\nPercentage with at least one future click:")
print(
    round(
        (model_df["future_clicks"] > 0).mean() * 100,
        2
    )
)
```

```python
model_df["target_has_clicks"] = (
    model_df["future_clicks"] > 0
).astype(int)

print("Target distribution:")
print(model_df["target_has_clicks"].value_counts())

print("\nTarget percentages:")
print(
    model_df["target_has_clicks"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
print("Available columns:")
print(model_df.columns.tolist())
```

```python

feature_columns = [
    "historical_impressions",
    "historical_clicks",
    "historical_avg_position",
    "historical_ctr",
    "impression_variability"
]

X = model_df[feature_columns].copy()
y = model_df["target_has_clicks"].copy()

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
print("Selected features:", X.columns.tolist())
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

model = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        max_iter=1000,
        random_state=42
    )
)

model.fit(X_train, y_train)

print("Model trained successfully!")
print("Training accuracy:", round(model.score(X_train, y_train), 4))
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

y_pred = model.predict(X_test)

print("Test accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred), 4))
print("Recall:", round(recall_score(y_test, y_pred), 4))
print("F1 score:", round(f1_score(y_test, y_pred), 4))

print("\nConfusion matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python

train_clients = set(
    model_df.loc[X_train.index, "client_hash_id"]
)

test_clients = set(
    model_df.loc[X_test.index, "client_hash_id"]
)

overlap = train_clients.intersection(test_clients)

print("Training clients:", len(train_clients))
print("Testing clients:", len(test_clients))
print("Clients in both sets:", len(overlap))
print(
    "Test client overlap (%):",
    round(len(overlap) / len(test_clients) * 100, 2)
)
```

```python

from sklearn.model_selection import GroupShuffleSplit

groups = model_df["client_hash_id"]

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train_group = X.iloc[train_idx]
X_test_group = X.iloc[test_idx]

y_train_group = y.iloc[train_idx]
y_test_group = y.iloc[test_idx]

train_group_clients = set(groups.iloc[train_idx])
test_group_clients = set(groups.iloc[test_idx])

print("Training rows:", len(X_train_group))
print("Testing rows:", len(X_test_group))

print("Training clients:", len(train_group_clients))
print("Testing clients:", len(test_group_clients))

print(
    "Overlapping clients:",
    len(train_group_clients & test_group_clients)
)

print("\nTraining target distribution:")
print(y_train_group.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test_group.value_counts(normalize=True).round(3))
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

group_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        max_iter=1000,
        random_state=42
    )
)

group_model.fit(X_train_group, y_train_group)

print("Client-aware model trained successfully!")

print(
    "Training accuracy:",
    round(group_model.score(X_train_group, y_train_group), 4)
)

print(
    "Testing accuracy:",
    round(group_model.score(X_test_group, y_test_group), 4)
)
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

y_pred_group = group_model.predict(X_test_group)

print("Client-aware model evaluation:")
print("Accuracy:", round(accuracy_score(y_test_group, y_pred_group), 4))
print("Precision:", round(precision_score(y_test_group, y_pred_group), 4))
print("Recall:", round(recall_score(y_test_group, y_pred_group), 4))
print("F1 score:", round(f1_score(y_test_group, y_pred_group), 4))

print("\nConfusion matrix:")
print(confusion_matrix(y_test_group, y_pred_group))
```

```python

import numpy as np
from sklearn.metrics import accuracy_score, f1_score

baseline_pred = np.zeros(len(y_test_group), dtype=int)

print("Baseline accuracy:",
      round(accuracy_score(y_test_group, baseline_pred), 4))

print("Baseline F1:",
      round(f1_score(y_test_group, baseline_pred), 4))

print("Client-aware model accuracy:",
      round(accuracy_score(y_test_group, y_pred_group), 4))

print("Client-aware model F1:",
      round(f1_score(y_test_group, y_pred_group), 4))
```

```python

from sklearn.metrics import roc_auc_score

y_prob_group = group_model.predict_proba(
    X_test_group
)[:, 1]

auc = roc_auc_score(y_test_group, y_prob_group)

print("Client-aware ROC-AUC:", round(auc, 4))

print(
    "Predicted probability range:",
    round(y_prob_group.min(), 4),
    "to",
    round(y_prob_group.max(), 4)
)
```

The dataset has several limitations. First, historical coverage is uneven across clients and pages, which may affect the reliability of the features. Second, early records rely on Google Search Console (GSC) data, so they cannot explain all factors influencing future clicks. Third, overlapping measurement windows may create dependencies between observations and potentially overestimate model performance.

Although the client-aware split prevents the same clients from appearing in both training and testing, the test set contains only nine clients. Therefore, the results may not generalize to every new client.

The model can identify statistical relationships between historical search performance and future clicks, but it cannot establish causation or guarantee future results.
