**GSC Data Availability Verification**

The March 2026 dataset contains 9,841,378 records.

Google Search Console (GSC) data is available for 3,611,061 records (36.69%) and unavailable for 6,230,317 records (63.31%).

This indicates that GSC coverage is limited. Records without available GSC data should not automatically be interpreted as having zero search activity.

The model must account for data availability when selecting features and interpreting results. This limitation may affect the generalizability of the findings.

```python

gsc_features = [
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position"
]

print("GSC feature missing-value check:")

for feature in gsc_features:
    missing = df[feature].isna().sum()
    percentage = df[feature].isna().mean() * 100

    print(
        f"{feature}: {missing:,} missing "
        f"({percentage:.2f}%)"
    )
```

**GSC Feature Completeness Verification**

The March 2026 dataset contains no null values in `gsc_impressions` or `gsc_clicks`.

However, `gsc_avg_position` has 6,230,317 missing values, representing 63.31% of records.

This missing-value count matches the number of records marked as having unavailable GSC data, suggesting that average-position completeness is related to GSC availability.

Although impressions and clicks contain no null values, this does not necessarily mean those measurements are available or valid for every record. Data availability flags must also be considered when interpreting these features.

```python

# Check GSC availability against missing average position
print("GSC availability vs missing average position:")

print(
    pd.crosstab(
        df["gsc_data_available"],
        df["gsc_avg_position"].isna(),
        rownames=["GSC available"],
        colnames=["Average position missing"]
    )
)

# Check whether the two conditions match for every row
matches = (
    (~df["gsc_data_available"]) ==
    df["gsc_avg_position"].isna()
)

print("\nAll records match:", matches.all())
print("Mismatched records:", (~matches).sum())
```

**GSC Availability and Missing-Value Consistency**

The cross-tabulation confirms that all 6,230,317 records marked as GSC unavailable have missing average-position values.

All 3,611,061 records marked as GSC available contain an average-position value.

The consistency check returned zero mismatched records across the entire dataset.

This demonstrates that missing average-position values align exactly with the GSC availability flag. These missing values should not automatically be replaced with zero, because unavailable measurements and actual zero values have different meanings.

```python
import pandas as pd

available_df = available_df.copy()

available_df["report_date"] = pd.to_datetime(
    available_df["report_date"]
)

print("Date type:", available_df["report_date"].dtype)
```

```python
history = available_df[
    available_df["report_date"].between(
        "2026-03-01", "2026-03-15"
    )
]

feature_df = (
    history.groupby(
        ["client_hash_id", "content_hash_id"]
    )
    .agg(
        historical_impressions=("gsc_impressions", "sum"),
        historical_clicks=("gsc_clicks", "sum"),
        historical_avg_position=("gsc_avg_position", "mean"),
        historical_ctr=("gsc_clicks", "sum"),
        impression_variability=("gsc_impressions", "std")
    )
    .reset_index()
)

feature_df["historical_ctr"] = (
    feature_df["historical_clicks"] /
    feature_df["historical_impressions"]
).fillna(0)

feature_df["impression_variability"] = (
    feature_df["impression_variability"].fillna(0)
)

print("Feature rows:", len(feature_df))
print(feature_df.head())
```

```python
import pandas as pd

available_df["report_date"] = pd.to_datetime(
    available_df["report_date"]
)

print("Date type:", available_df["report_date"].dtype)
```

```python
print("Feature rows:", len(feature_df))

print("\nMissing values:")
print(feature_df.isna().sum())

print("\nDuplicate client-page combinations:")
print(
    feature_df.duplicated(
        subset=["client_hash_id", "content_hash_id"]
    ).sum()
)
```

```python
print("History start:", history["report_date"].min())
print("History end:", history["report_date"].max())
print("Historical rows:", len(history))
print("Feature rows:", len(feature_df))
```

```python
future = available_df[
    available_df["report_date"].between(
        "2026-03-16", "2026-03-31"
    )
].copy()

print("Future start:", future["report_date"].min())
print("Future end:", future["report_date"].max())
print("Future rows:", len(future))
print("Unique client-page pairs:",
      future[["client_hash_id", "content_hash_id"]]
      .drop_duplicates().shape[0])
```

```python
future_pairs = future[
    ["client_hash_id", "content_hash_id"]
].drop_duplicates()

matched_pairs = feature_df.merge(
    future_pairs,
    on=["client_hash_id", "content_hash_id"],
    how="inner"
)

print("Historical feature pairs:", len(feature_df))
print("Future unique pairs:", len(future_pairs))
print("Matched pairs:", len(matched_pairs))

print(
    "Historical coverage (%):",
    round(len(matched_pairs) / len(feature_df) * 100, 2)
)
```

```python
future_targets = (
    future.groupby(
        ["client_hash_id", "content_hash_id"]
    )
    .agg(
        future_impressions=("gsc_impressions", "sum"),
        future_clicks=("gsc_clicks", "sum")
    )
    .reset_index()
)

print("Future target rows:", len(future_targets))

print("\nMissing values:")
print(future_targets.isna().sum())

print("\nFirst five rows:")
print(future_targets.head())
```

```python
model_df = feature_df.merge(
    future_targets,
    on=["client_hash_id", "content_hash_id"],
    how="inner",
    validate="one_to_one"
)

print("Model dataset rows:", len(model_df))

print("\nMissing values:")
print(model_df.isna().sum())

print("\nDuplicate client-page pairs:")
print(
    model_df.duplicated(
        subset=["client_hash_id", "content_hash_id"]
    ).sum()
)

print("\nFirst five rows:")
print(model_df.head())
```

```python
print("Future clicks summary:")
print(model_df["future_clicks"].describe())

print("\nPercentage with zero future clicks:")
print(
    round(
        (model_df["future_clicks"] == 0).mean() * 100,
        2
    )
)

print("\nPercentage with at least one future click:")
print(
    round(
        (model_df["future_clicks"] > 0).mean() * 100,
        2
    )
)
```

```python
model_df["target_has_clicks"] = (
    model_df["future_clicks"] > 0
).astype(int)

print("Target distribution:")
print(model_df["target_has_clicks"].value_counts())

print("\nTarget percentages:")
print(
    model_df["target_has_clicks"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
print("Available columns:")
print(model_df.columns.tolist())
```

```python

feature_columns = [
    "historical_impressions",
    "historical_clicks",
    "historical_avg_position",
    "historical_ctr",
    "impression_variability"
]

X = model_df[feature_columns].copy()
y = model_df["target_has_clicks"].copy()

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
print("Selected features:", X.columns.tolist())
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

model = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        max_iter=1000,
        random_state=42
    )
)

model.fit(X_train, y_train)

print("Model trained successfully!")
print("Training accuracy:", round(model.score(X_train, y_train), 4))
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

y_pred = model.predict(X_test)

print("Test accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred), 4))
print("Recall:", round(recall_score(y_test, y_pred), 4))
print("F1 score:", round(f1_score(y_test, y_pred), 4))

print("\nConfusion matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python

train_clients = set(
    model_df.loc[X_train.index, "client_hash_id"]
)

test_clients = set(
    model_df.loc[X_test.index, "client_hash_id"]
)

overlap = train_clients.intersection(test_clients)

print("Training clients:", len(train_clients))
print("Testing clients:", len(test_clients))
print("Clients in both sets:", len(overlap))
print(
    "Test client overlap (%):",
    round(len(overlap) / len(test_clients) * 100, 2)
)
```

```python

from sklearn.model_selection import GroupShuffleSplit

groups = model_df["client_hash_id"]

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train_group = X.iloc[train_idx]
X_test_group = X.iloc[test_idx]

y_train_group = y.iloc[train_idx]
y_test_group = y.iloc[test_idx]

train_group_clients = set(groups.iloc[train_idx])
test_group_clients = set(groups.iloc[test_idx])

print("Training rows:", len(X_train_group))
print("Testing rows:", len(X_test_group))

print("Training clients:", len(train_group_clients))
print("Testing clients:", len(test_group_clients))

print(
    "Overlapping clients:",
    len(train_group_clients & test_group_clients)
)

print("\nTraining target distribution:")
print(y_train_group.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test_group.value_counts(normalize=True).round(3))
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

group_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        max_iter=1000,
        random_state=42
    )
)

group_model.fit(X_train_group, y_train_group)

print("Client-aware model trained successfully!")

print(
    "Training accuracy:",
    round(group_model.score(X_train_group, y_train_group), 4)
)

print(
    "Testing accuracy:",
    round(group_model.score(X_test_group, y_test_group), 4)
)
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

y_pred_group = group_model.predict(X_test_group)

print("Client-aware model evaluation:")
print("Accuracy:", round(accuracy_score(y_test_group, y_pred_group), 4))
print("Precision:", round(precision_score(y_test_group, y_pred_group), 4))
print("Recall:", round(recall_score(y_test_group, y_pred_group), 4))
print("F1 score:", round(f1_score(y_test_group, y_pred_group), 4))

print("\nConfusion matrix:")
print(confusion_matrix(y_test_group, y_pred_group))
```
