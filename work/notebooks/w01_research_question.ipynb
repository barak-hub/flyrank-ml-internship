### Quick Look at the Data

The FlyRank starter dataset contains **30,000 content items**. My analysis found that **19,716 items (65.72%)** had fewer Google Search impressions in the most recent 30-day period compared with the preceding 30-day period.

These results suggest that declining search visibility is common in this dataset and provide a reason to investigate which pages may benefit from SEO optimization.

This supports my choice of the **Google Search Ranking & Discoverability** research lane. Machine learning could help identify patterns associated with declining performance and support the prioritization of pages for further investigation.

However, declining impressions alone do not prove that a page needs optimization or that any particular action will improve its ranking.

```python

import pandas as pd

url = "https://raw.githubusercontent.com/barak-hub/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Dataset rows:", len(df))
print("Dataset columns:", len(df.columns))
print("Column names:", df.columns.tolist())
```

```python

for column in df.columns:
    print(column)
```

```python

print("Total content items:", len(df))

declining = (
    df["impressions_last_30d"]
    < df["impressions_prev_30d"]
)

print("Pages with declining impressions:", declining.sum())

print(
    "Percentage with declining impressions:",
    round(declining.mean() * 100, 2),
    "%"
)
```
